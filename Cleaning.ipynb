```python
# num_rear_cameras find function.

def camera_extractor(text):

    if 'Quad' in text:
        return 4
    elif 'Triple' in text:
        return 3
    elif 'Dual' in text:
        return 2
    else:
        return 1
```

```python
num_rear_cameras = df1['camera'].str.split('&').str.get(0).str.strip().apply(camera_extractor)
num_rear_cameras
```

```python
df1.insert(25,'num_rear_cameras',num_rear_cameras)
```

```python
 # Find the 
def camera_extractor2(text):

    if 'Quad' in text:
        return '4'
    elif 'Triple' in text:
        return '3'
    elif 'Dual' in text:
        return '2'
    elif 'Missing' in text:
        return 'Missing'
    else:
        return '1'
```

```python
num_front_cameras = df1['camera'].str.split('&').str.get(1).str.strip().fillna('Missing').apply(camera_extractor2)
num_front_cameras
```
